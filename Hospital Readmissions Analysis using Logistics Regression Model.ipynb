# Hospital Readmissions Analysis Using Logistic Regression
The objective of this project is to analyze and predict hospital readmissions by evaluating patient data such as diagnoses, medication use, length of stay, and treatment history. The dataset undergoes comprehensive preprocessing, including handling missing values, encoding categorical features, and standardizing numerical variables to enhance model performance. Logistic Regression serves as the core modeling approach, with evaluation metrics such as accuracy, precision, recall, and AUC-ROC used to measure its effectiveness. Hyperparameter tuning is conducted using tools like GridSearchCV to improve model robustness. This project aims to support healthcare providers in identifying at-risk patients, optimizing treatment plans, and reducing unnecessary hospital readmissions for better resource management and patient care.

## 📈 4. Model Training – Logistic Regression

```python
x = df.drop("Readmitted", axis=1)  # Replace 'Adulteration' with your actual target column name
y = df["Readmitted"]
```

```python
y
```

```python
for col in x.select_dtypes(include='object').columns:
    le = LabelEncoder()
    x[col] = le.fit_transform(x[col])
```

## 🧪 5. Train-Test Split

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
log_reg = LogisticRegression()
log_reg.fit(x_train, y_train)
```

```python
LOR = LogisticRegression()
LOR
```

```python
LOR.fit(x_train, y_train)
```

## ✅ 6. Model Evaluation

```python
y_pred= LOR.predict(x_test)
```

```python
classification = classification_report(y_test, y_pred)
```

```python
acc =accuracy_score(y_test,y_pred)
acc
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

#### **Hyperparameter Tuning**

```python
param_grid = {
    'C': [0.01, 0.1, 1, 10],
    'penalty': ['l1', 'l2'],
    'solver': ['liblinear']
}

grid = GridSearchCV(LogisticRegression(max_iter=1000), param_grid, cv=5)
grid.fit(x_train, y_train)

print("Best Parameters:", grid.best_params_)
```

#### **Retrain with Best Params**

```python
best_model = grid.best_estimator_
y_pred_optimized = best_model.predict(x_test)

print("Optimized Accuracy:", accuracy_score(y_test, y_pred_optimized))
```
